## Step 2: Ingesting Financial Market History

### 1. Purpose & Core Objective
Load historical market candles from `data/stock_market/` and compute technical indicators.

### 2. Real-World Analogy & Beginner Intuition
Loading the Bloomberg terminal database with historical daily stock prices and volume records.

### 3. Inputs & Flow from Previous Step
- **Input Variables / Artifacts Taken**: `load_dataset` helper from Step 1.
- **Why We Need Them Now**: They provide the clean foundation required to perform this next transformation.

### 4. What This Step Does
Loads DataFrame `df`, identifies price columns, and computes 20-day / 50-day moving averages and 14-day RSI.

### 5. What It Will Be Used For
Provides the factual market database queried by the agent swarm.


```python
df = load_dataset('stock_market')
close_col = [c for c in df.columns if 'close' in c.lower()][0]
prices = df[close_col].dropna().values

# Compute Technical Indicators: 20-Day SMA, 50-Day SMA, RSI
sma_20 = pd.Series(prices).rolling(20).mean().values
sma_50 = pd.Series(prices).rolling(50).mean().values

# 14-Day Relative Strength Index (RSI)
deltas = np.diff(prices)
gains = np.where(deltas > 0, deltas, 0)
losses = np.where(deltas < 0, -deltas, 0)
avg_gain = pd.Series(gains).rolling(14).mean().values
avg_loss = pd.Series(losses).rolling(14).mean().values
rs = np.divide(avg_gain, avg_loss, out=np.zeros_like(avg_gain), where=avg_loss!=0)
rsi = 100 - (100 / (1 + rs))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

# 1. Price and MAs
axes[0].plot(prices[-150:], label='Close Price ($)', color='#2c3e50', lw=1.5)
axes[0].plot(sma_20[-150:], label='20-Day SMA', color='#e74c3c', lw=1.2)
axes[0].plot(sma_50[-150:], label='50-Day SMA', color='#3498db', lw=1.2)
axes[0].set_title("Historical Price & Moving Averages (Last 150 Days)", fontsize=12, fontweight='bold')
axes[0].set_xlabel('Trading Day', fontsize=10)
axes[0].set_ylabel('Price ($)', fontsize=10)
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.5)

# 2. RSI Indicator
axes[1].plot(rsi[-150:], color='#8e44ad', lw=1.5)
axes[1].axhline(70, color='red', linestyle='--', label='Overbought (70)')
axes[1].axhline(30, color='green', linestyle='--', label='Oversold (30)')
axes[1].set_title(f"14-Day RSI (Latest: {rsi[-1]:.1f})", fontsize=12, fontweight='bold')
axes[1].set_xlabel('Trading Day', fontsize=10)
axes[1].set_ylabel('RSI Score (0-100)', fontsize=10)
axes[1].legend()
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()


```

## Step 3: Defining Executable Financial Tools with Structured Schemas

### 1. Purpose & Core Objective
Implement deterministic Python tools for price lookup, momentum calculation, and portfolio risk evaluation.

### 2. Real-World Analogy & Beginner Intuition
Giving our AI analysts financial formulas they can run directly instead of guessing stock math.

### 3. Inputs & Flow from Previous Step
- **Input Variables / Artifacts Taken**: Market data arrays from Step 2.
- **Why We Need Them Now**: They provide the clean foundation required to perform this next transformation.

### 4. What This Step Does
Defines `tool_get_price`, `tool_get_rsi`, and `tool_calculate_var` functions with JSON schemas.

### 5. What It Will Be Used For
Provides the executable actions dispatched by the agent swarm.


```python
def tool_get_price(ticker: str = "ASSET") -> dict:
    return {
        "status": "success",
        "ticker": ticker,
        "latest_close": round(float(prices[-1]), 2),
        "52w_high": round(float(np.max(prices[-252:])), 2),
        "52w_low": round(float(np.min(prices[-252:])), 2)
    }

def tool_get_rsi(ticker: str = "ASSET") -> dict:
    val = float(rsi[-1])
    regime = "Overbought" if val > 70 else ("Oversold" if val < 30 else "Neutral")
    return {
        "status": "success",
        "rsi_14d": round(val, 2),
        "regime": regime
    }

def tool_calculate_var(ticker: str = "ASSET", confidence: float = 0.95) -> dict:
    pct_returns = np.diff(prices[-252:]) / prices[-252:-1]
    var_pct = np.percentile(pct_returns, (1 - confidence) * 100)
    return {
        "status": "success",
        "confidence": confidence,
        "daily_var_pct": round(float(var_pct) * 100, 2),
        "interpretation": f"95% of daily losses do not exceed {abs(var_pct)*100:.2f}%"
    }

FINANCIAL_TOOLS = {
    "get_price": tool_get_price,
    "get_rsi": tool_get_rsi,
    "calculate_var": tool_calculate_var
}

print("Registered Financial Analyst Tools:")
for k in FINANCIAL_TOOLS:
    print(f"- Tool: `{k}`")


```

#### 1. Metric & Value Breakdown
- **Deterministic Tools**: 3 financial calculation tools registered with structured outputs.

#### 3. Key Takeaway & Next Action

## Step 4: Multi-Agent Swarm Orchestrator & Synthesis

### 1. Purpose & Core Objective
Implement an orchestrator coordinating three specialized agents (Technical, Valuation, Risk) to produce a unified investment recommendation.

### 2. Real-World Analogy & Beginner Intuition
A Wall Street investment committee meeting: the Technical Analyst presents trend charts, the Valuation Analyst checks pricing, and the Risk Officer checks downside limits before signing off on a trade.

### 3. Inputs & Flow from Previous Step
- **Input Variables / Artifacts Taken**: `FINANCIAL_TOOLS` from Step 3.
- **Why We Need Them Now**: They provide the clean foundation required to perform this next transformation.

### 4. What This Step Does
Simulates multi-agent reasoning, executes tool calls, and synthesizes a structured investment memo.

### 5. What It Will Be Used For
Forms the production multi-agent engine.


```python
class MarketAnalystSwarm:
    def __init__(self, tools):
        self.tools = tools
        
    def generate_research_report(self, ticker: str) -> dict:
        print(f" Swarm Initiating Analysis on Ticker: {ticker}")
        
        # Agent 1: Technical Analyst
        rsi_data = self.tools['get_rsi'](ticker)
        print(f"-> Technical Analyst Dispatched `get_rsi`: {rsi_data}")
        
        # Agent 2: Valuation Analyst
        price_data = self.tools['get_price'](ticker)
        print(f"-> Valuation Analyst Dispatched `get_price`: {price_data}")
        
        # Agent 3: Risk Officer
        risk_data = self.tools['calculate_var'](ticker)
        print(f"-> Risk Officer Dispatched `calculate_var`: {risk_data}")
        
        # Swarm Synthesis
        recommendation = "ACCUMULATE (BUY)" if rsi_data['rsi_14d'] < 65 else "HOLD / TRIM"
        
        report = {
            "ticker": ticker,
            "latest_price": price_data['latest_close'],
            "technical_regime": rsi_data['regime'],
            "rsi": rsi_data['rsi_14d'],
            "daily_var_95": f"{risk_data['daily_var_pct']}%",
            "consensus_recommendation": recommendation,
            "executive_memo": f"Asset {ticker} trades at ${price_data['latest_close']:.2f} with RSI {rsi_data['rsi_14d']:.1f} ({rsi_data['regime']}). Daily 95% VaR is capped at {risk_data['daily_var_pct']}%. Swarm consensus is {recommendation}."
        }
        return report

swarm = MarketAnalystSwarm(FINANCIAL_TOOLS)
report = swarm.generate_research_report("AAPL")
print(" Swarm Investment Research Memo:")
print(json.dumps(report, indent=2))


```

#### 1. Metric & Value Breakdown
- **Multi-Agent Consensus**: All three agents successfully executed tool calls and converged on a unified research report.
