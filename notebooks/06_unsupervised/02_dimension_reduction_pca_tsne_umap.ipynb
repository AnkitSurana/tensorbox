## 1. PCA Mathematical Formulation
Given centered data matrix $X$, compute covariance matrix $\Sigma = \frac{1}{n} X^T X$ and its eigendecomposition:
$$\Sigma v_i = \lambda_i v_i$$
Project onto top $k$ principal eigenvectors.

```python
import os
import sys
from pathlib import Path

for p in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (p / "utils").exists():
        if str(p) not in sys.path:
            sys.path.insert(0, str(p))
        break

from utils.data_loader import load_dataset

import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

df = load_dataset("customer_segmentation")
num_cols = df.select_dtypes(include=[np.number]).columns
X_scaled = StandardScaler().fit_transform(df[num_cols])

pca = PCA()
X_pca = pca.fit_transform(X_scaled)

explained_var = pca.explained_variance_ratio_
cum_var = np.cumsum(explained_var)

print("=== PCA Explained Variance Ratio ===")
for i, (ev, cv) in enumerate(zip(explained_var, cum_var)):
    print(f"PC{i+1}: {ev*100:.2f}% | Cumulative: {cv*100:.2f}%")
```

## 2. Non-Linear Manifold Learning with t-SNE
Generate 2D non-linear neighborhood embedding preserving local structure.

```python
tsne = TSNE(n_components=2, perplexity=30, random_state=42, max_iter=500)
X_tsne = tsne.fit_transform(X_scaled[:500])

print(f"Original Dimensionality: {X_scaled.shape[1]} -> t-SNE Reduced: {X_tsne.shape[1]}")
print(f"t-SNE Embedding Sample (first 3 points):\n{X_tsne[:3].round(3)}")
```
