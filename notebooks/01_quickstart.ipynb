## 2. Fast Neural Embeddings & Vector Search (ChromaDB)
ChromaDB is pre-installed. Let's create a local collection and query it using semantic similarity.

```python
import chromadb
from pathlib import Path

models_dir = Path("./models/chromadb")
models_dir.mkdir(parents=True, exist_ok=True)

client = chromadb.PersistentClient(path=str(models_dir))
collection = client.get_or_create_collection("quickstart_docs")

docs = [
    "Tensorbox is a zero-setup machine learning workspace.",
    "PyTorch and TensorFlow provide deep learning foundations.",
    "FastAPI powers high-performance model serving endpoints."
]
ids = ["doc_1", "doc_2", "doc_3"]

collection.upsert(documents=docs, ids=ids)
results = collection.query(query_texts=["machine learning tools"], n_results=2)

print("=== Chroma Vector Search Results ===")
for doc, doc_id in zip(results["documents"][0], results["ids"][0]):
    print(f"[{doc_id}]: {doc}")
```

## 3. High-Speed Dense Vector Search (FAISS)
Using FAISS with Sentence Transformers for in-memory similarity matching.

```python
import faiss
from sentence_transformers import SentenceTransformer

# Load sentence transformer model
embedder = SentenceTransformer('all-MiniLM-L6-v2')

corpus = [
    "PyTorch provides tensor computation and deep neural networks on GPUs.",
    "Scikit-learn is standard for classical machine learning algorithms and preprocessing.",
    "FastAPI delivers high-performance REST APIs with automatic OpenAPI Swagger docs."
]

embeddings = embedder.encode(corpus).astype('float32')
dimension = embeddings.shape[1]

# Build FAISS index
index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

# Query
query = "How do I build machine learning REST web services?"
query_vec = embedder.encode([query]).astype('float32')
distances, indices = index.search(query_vec, k=1)

print(f"Query: {query}")
print(f"Best Match: {corpus[indices[0][0]]} (L2 Distance: {distances[0][0]:.4f})")
```
